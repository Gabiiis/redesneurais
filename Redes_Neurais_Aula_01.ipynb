### Exemplo de `while` percorrendo uma lista

No `while`, controlamos manualmente o índice da lista de `0` até o tamanho máximo dela.

```python
indice = 0
tamanho_lista = len(notas)

print("Percorrendo a lista de notas com while:")
while indice < tamanho_lista:
    print(f"Nota no índice {indice}: {notas[indice]}")
    indice += 1 # Incrementa o índice para ir para o próximo elemento
```

```python
import pandas as pd
```

```python
dataset = pd.DataFrame({
    "Produto": ["A", "B", "C", "D"],
    "Preço": [20, 40, 60, np.nan],
    "Quantidade": [10, 20, 30, 40]
})
print(dataset)
```

```python
dataset.describe()
```

```python
dataset.info()
```

```python
mediana_preco = dataset["Preço"].median() # chamando método da biblioteca para calcular a mediana
dataset['Preço'] = dataset['Preço'].fillna(mediana_preco) # substitui valores ausentes pela mediana
print(dataset)
```

```python
dataset.describe()
```

```python
# Agrupando o dataset por Produto para analisar o Preço e Quantidade acumulados
dataset_agrupado = dataset.groupby('Produto').agg({
    'Preço': 'sum',
    'Quantidade': 'sum'
}).reset_index()

display(dataset_agrupado)
```

### Usando o método `apply` no Pandas

Podemos usar o `.apply()` para criar novas colunas com base em funções personalizadas. No exemplo abaixo, criamos uma função para classificar a faixa de preço do produto.

```python
# Definindo uma função personalizada para classificar a faixa de preço
def categorizar_preco(linha):
    if linha['Preço'] > 150:
        return 'Alto'
    else:
        return 'Baixo'

# Aplicando a função ao dataset_agrupado
# Usamos axis=1 para indicar que queremos aplicar a função linha por linha
dataset_agrupado['Categoria_Preço'] = dataset_agrupado.apply(categorizar_preco, axis=1)

display(dataset_agrupado)
```

```python
produtos_por_preco = dataset.groupby('Preço')['Produto'].apply(list).reset_index()
print(produtos_por_preco)
```

```python
df_filtrado = dataset[dataset['Preço'] > 30]
print(df_filtrado)
```

### Visualização de Dados: Histograma de Preços

Vamos criar um histograma simples para visualizar a distribuição dos preços no nosso dataset.

```python
# Criando o histograma da coluna 'Preço'
plt.figure(figsize=(6, 4))
sns.histplot(dataset['Preço'], kde=True, bins=3)

# Adicionando títulos e rótulos
#plt.title('Distribuição de Preços dos Produtos')
plt.xlabel('Preço')
plt.ylabel('Frequência')

# Exibindo o gráfico

plt.savefig('histograma_precos.png')
plt.show()
```

### Remoção de colunas com alta quantidade de dados faltantes
